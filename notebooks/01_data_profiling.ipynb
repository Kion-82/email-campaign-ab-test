```python
import pandas as pd

df = pd.read_csv("../data/raw/hillstrom.csv")
print(df.shape)
df.head()
```

```python
df.info()
print("\nMissing values per column:")
print(df.isna().sum())
```

```python
df[["recency", "history", "spend"]].describe()
```

```python
low_cardinality_cols = [
    "history_segment", "zip_code", "channel", "segment",
    "mens", "womens", "newbie", "visit", "conversion",
]
for col in low_cardinality_cols:
    print(df[col].value_counts(dropna=False).sort_index(), "\n")
```

```python
# Longest string in each text column, used to size VARCHAR columns
print("Max string length:")
print(df.select_dtypes("object").apply(lambda s: s.str.len().max()))

# Maximum number of decimal places in the monetary columns, used to choose DECIMAL scale
for col in ["history", "spend"]:
    decimals = df[col].astype(str).str.split(".").str[1].str.len().max()
    print(f"{col}: max {decimals} decimal places, max value {df[col].max()}")
```
